IT8 — Fine: Certify the Machine


```python
class SpecViolation(Exception):
    """Raised when a job violates machine specifications."""
    pass


class Machine:

    def __init__(self, name, capacity, rating, limit):
        self.name = name
        self.capacity = capacity
        self.rating = rating
        self.limit = limit

        self.__jobs_completed = 0
        self.__total_processed = 0

    def operate(self, job):
        """Processes a job after validating specifications."""

        if not isinstance(job, (int, float)):
            raise SpecViolation("Job must be numeric")

        if job <= 0:
            raise SpecViolation("Job must be positive")

        if job > self.limit:
            raise SpecViolation("Job exceeds machine limit")

        self.__jobs_completed += 1
        self.__total_processed += job

        return f"{self.name} processed job {job}"

    def status(self):
        """Returns machine statistics."""

        return {
            "Name": self.name,
            "Jobs Completed": self.__jobs_completed,
            "Total Processed": self.__total_processed
        }


# Create machines
machine1 = Machine("Machine A", 100, 5, 50)
machine2 = Machine("Machine B", 150, 4, 70)
machine3 = Machine("Machine C", 200, 5, 100)

machines = [machine1, machine2, machine3]

# Jobs to process
jobs = [20, 40, 150, 30, 60, 10, -5, 80]

# Process jobs without stopping the fleet
for job in jobs:

    for machine in machines:

        try:
            result = machine.operate(job)
            print(result)
            break

        except SpecViolation as error:
            print(f"{machine.name} rejected job {job}: {error}")

    else:
        print(f"Job {job} rejected by all machines")


# Final report
print("\nFLEET REPORT")
print("-------------------------")

for machine in machines:
    print(machine.status())
```

IT6 — Precision: The Self-Testing Program

```python
class SpecViolation(Exception):
    """Raised when machine specifications are violated."""
    pass


class Machine:

    def __init__(self, name, capacity, rating, limit):
        self.name = name
        self.capacity = capacity
        self.rating = rating
        self.limit = limit

        self.__jobs_completed = 0
        self.__total_processed = 0

    def operate(self, job):
        """Validates and processes a machine job."""

        if not isinstance(job, (int, float)):
            raise SpecViolation("Job must be numeric")

        if job <= 0:
            raise SpecViolation("Job must be positive")

        if job > self.limit:
            raise SpecViolation("Job exceeds machine limit")

        self.__jobs_completed += 1
        self.__total_processed += job

        return True

    def status(self):
        """Returns machine statistics."""

        return {
            "Name": self.name,
            "Jobs Completed": self.__jobs_completed,
            "Total Processed": self.__total_processed
        }


def parse_line(line):
    """Parses and validates a parts log line."""

    if not isinstance(line, str):
        raise TypeError("Input must be a string")

    parts = line.strip().split(",")

    if len(parts) != 2:
        raise ValueError("Expected two values")

    code = parts[0].strip()

    if not code:
        raise ValueError("Part code is empty")

    diameter = float(parts[1].strip())

    if diameter <= 0:
        raise ValueError("Diameter must be positive")

    if diameter > 1000:
        raise ValueError("Diameter too large")

    return {
        "code": code,
        "diameter": diameter
    }


def run_tests():
    """Runs eight hostile parser tests and prints pass/fail results."""

    hostile = [
        "",
        "INVALID DATA",
        "101,25",
        "A101,abc",
        "A101,-10",
        "A101,0",
        "A101,1000000",
        None
    ]

    print("\nSELF TEST REPORT")
    print("=" * 50)

    passed = 0
    failed = 0

    for number, line in enumerate(hostile, start=1):

        try:
            parse_line(line)

        except (TypeError, ValueError):
            print(f"Test {number}: PASS")
            passed += 1

        else:
            print(f"Test {number}: FAIL")
            failed += 1

    print("=" * 50)
    print("Total Tests:", len(hostile))
    print("Passed:", passed)
    print("Failed:", failed)

    return failed == 0
```
