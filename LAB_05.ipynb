IT12 – One Line, Three Fields
Extract the part code, diameter and shift from one raw string.

```python
line = "P101, 25.02, DAY"

parts = line.split(",")

code = parts[0].strip()
dia = float(parts[1].strip())
shift = parts[2].strip()

print("Part Code:", code)
print("Diameter:", dia)
print("Shift:", shift)

print("Diameter + 0.01:", dia + 0.01)
```

IT8 – The Shift Report
Now sort all 10 diameters into ACCEPT, REWORK and SCRAP lists.

```python
lines = [
    "P101, 25.02, DAY",
    " P102 , 24.98 , DAY ",
    "P103,25.07,DAY",
    " P104, 25.12, DAY",
    "P105 , 24.93 , DAY",
    "P106, 25.00, DAY",
    " P107,25.04,DAY",
    "P108 , 25.09 , DAY",
    "P109, 24.89, DAY",
    " P110 , 25.01 , DAY"
]

accepted = []
rework = []
scrap = []
scrap_codes = []

for line in lines:
    parts = line.split(",")

    code = parts[0].strip()
    dia = float(parts[1].strip())

    if 24.95 <= dia <= 25.05:
        accepted.append(dia)

    elif 24.90 <= dia <= 25.10:
        rework.append(dia)

    else:
        scrap.append(dia)
        scrap_codes.append(code)

print("ACCEPT count:", len(accepted))
print("REWORK count:", len(rework))
print("SCRAP count:", len(scrap))

if len(accepted) > 0:
    print("Accepted average:", sum(accepted) / len(accepted))
else:
    print("No accepted parts")

print("Scrapped part codes:", scrap_codes)
```

IT6 – Two Shifts Compared
This final program compares DAY and NIGHT shifts using lists and sets.

```python
# DAY SHIFT
day_lines = [
    "P101, 25.02, DAY",
    "P102, 25.12, DAY",
    "P103, 24.98, DAY",
    "P104, 25.15, DAY",
    "P105, 25.00, DAY"
]

# NIGHT SHIFT
night_lines = [
    "P101, 25.14, NIGHT",
    "P102, 25.01, NIGHT",
    "P106, 24.88, NIGHT",
    "P107, 25.12, NIGHT",
    "P108, 25.02, NIGHT"
]

day_scrap = []
night_scrap = []

day_accept = 0
night_accept = 0

# Process DAY SHIFT
for line in day_lines:
    parts = line.split(",")

    code = parts[0].strip()
    dia = float(parts[1].strip())

    if 24.95 <= dia <= 25.05:
        day_accept += 1
    elif 24.90 <= dia <= 25.10:
        pass
    else:
        day_scrap.append(code)

# Process NIGHT SHIFT
for line in night_lines:
    parts = line.split(",")

    code = parts[0].strip()
    dia = float(parts[1].strip())

    if 24.95 <= dia <= 25.05:
        night_accept += 1
    elif 24.90 <= dia <= 25.10:
        pass
    else:
        night_scrap.append(code)

# Calculate accept rates
day_rate = (day_accept / len(day_lines)) * 100
night_rate = (night_accept / len(night_lines)) * 100

# Compare scrap codes using sets
day_set = set(day_scrap)
night_set = set(night_scrap)

common = day_set & night_set
unique = day_set ^ night_set

# Final report
print("===== TWO SHIFT REPORT =====")

print("DAY accept rate:", day_rate, "%")
print("NIGHT accept rate:", night_rate, "%")

print("DAY scrap codes:", day_scrap)
print("NIGHT scrap codes:", night_scrap)

print("Scrap codes in BOTH shifts:", common)
print("Scrap codes unique to ONE shift:", unique)
```
