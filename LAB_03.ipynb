IT10 – The Batch Inspector
This program reads 10 diameters and sorts them into three bins.

```python
accept = 0
rework = 0
scrap = 0

for i in range(10):
    diameter = float(input("Enter diameter: "))

    if 9.5 <= diameter <= 10.5:
        print("ACCEPT")
        accept += 1

    elif 9.0 <= diameter <= 11.0:
        print("REWORK")
        rework += 1

    else:
        print("SCRAP")
        scrap += 1

print("Final Report")
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

IT6 – Precision: End of Shift

```python
# Batch Inspection

accept = 0
rework = 0
scrap = 0

for i in range(10):
    diameter = float(input("Enter part diameter: "))

    if 9.5 <= diameter <= 10.5:
        print("ACCEPT")
        accept += 1

    elif 9.0 <= diameter <= 11.0:
        print("REWORK")
        rework += 1

    else:
        print("SCRAP")
        scrap += 1

# Pack only accepted parts into boxes of 4

boxes = accept // 4
loose = accept % 4

# End of Shift Summary

print("\n===== END OF SHIFT SUMMARY =====")

print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)

print("Full boxes:", boxes)
print("Loose parts:", loose)

print("Audit:", boxes * 4 + loose == accept)
```
