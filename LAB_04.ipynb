IT12 – Ten From the Machine
Generate and print 10 random diameters.

```python
import random

for i in range(10):
    diameter = random.uniform(24.90, 25.10)
    print(round(diameter, 2))
```

IT10 – The Self-Sorting Bench
Generate 10 diameters, sort them into three bins and count each category.

```python
import random

accept = 0
rework = 0
scrap = 0

for i in range(10):
    diameter = round(random.uniform(24.90, 25.10), 2)

    if 24.95 <= diameter <= 25.05:
        print(diameter, "ACCEPT")
        accept += 1

    elif 24.90 <= diameter <= 25.10:
        print(diameter, "REWORK")
        rework += 1

    else:
        print(diameter, "SCRAP")
        scrap += 1

print("\n--- SHIFT REPORT ---")
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

IT8 – Halt on Scrap
The operator chooses between two modes:
- Halt: Stop immediately when a SCRAP part appears.
- Skip: Record the SCRAP part and continue inspecting.

```python
import random

mode = input("Choose mode (halt/skip): ").lower()

accept = 0
rework = 0
scrap = 0
inspected = 0

for i in range(100):

    diameter = round(random.uniform(24.90, 25.10), 2)
    inspected += 1

    if 24.95 <= diameter <= 25.05:
        print(diameter, "ACCEPT")
        accept += 1

    elif 24.90 <= diameter <= 25.10:
        print(diameter, "REWORK")
        rework += 1

    else:
        print(diameter, "SCRAP")
        scrap += 1

        if mode == "halt":
            print("SCRAP FOUND! MACHINE STOPPED.")
            break

        elif mode == "skip":
            continue

print("\n--- SHIFT REPORT ---")
print("Inspected:", inspected)
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

IT6 – Run Until Good
The machine keeps generating diameters until it achieves five consecutive ACCEPT parts.

```python
import random

accept = 0
rework = 0
scrap = 0

consecutive = 0
attempts = 0
max_attempts = 500

while consecutive < 5 and attempts < max_attempts:

    diameter = round(random.uniform(24.90, 25.10), 2)
    attempts += 1

    if 24.95 <= diameter <= 25.05:
        print(diameter, "ACCEPT")
        accept += 1
        consecutive += 1

    elif 24.90 <= diameter <= 25.10:
        print(diameter, "REWORK")
        rework += 1
        consecutive = 0

    else:
        print(diameter, "SCRAP")
        scrap += 1
        consecutive = 0

print("\n--- FINAL SHIFT REPORT ---")
print("Total attempts:", attempts)
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)

if consecutive == 5:
    print("SUCCESS: Five consecutive ACCEPTS achieved!")

else:
    print("SAFETY STOP: Maximum attempts reached.")
```
